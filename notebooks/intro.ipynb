# Statistics Lessons Demo
This notebook shows how to use utilities from the repository.

```python
import pandas as pd
from sklearn.datasets import load_iris
from foundations.descriptive_stats import summary_statistics
from ml_models.regression_examples import logistic_regression_example

# Load iris dataset
iris = load_iris(as_frame=True)
X = iris.data['sepal length (cm)']
y = (iris.target == 0).astype(int)

stats = summary_statistics(X)
print(stats)

coef, intercept = logistic_regression_example(X.values, y.values)
print(coef, intercept)
```
